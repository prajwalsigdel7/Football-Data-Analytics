```python
import understatapi
from understatapi import UnderstatClient
import pandas as pd
import matplotlib
from matplotlib import pyplot as plt
```

```python
# 1. Open the connection
with UnderstatClient() as understat:
    # Understat uses the start year (2025) for the 2025/26 season
    # We get player data for the whole league first
    league_data = understat.league(league="La_Liga").get_player_data(season="2025")

# 2. Convert to DataFrame
df = pd.DataFrame(league_data)

# 3. Filter for only Real Madrid players
# (Understat usually uses "Real Madrid" as the team_title)
real_madrid_df = df[df['team_title'] == 'Real Madrid'].copy()

# 4. Clean the data (Convert strings to numbers)
cols_to_fix = ['goals', 'assists', 'xG', 'xA', 'shots', 'key_passes']
for col in cols_to_fix:
    real_madrid_df[col] = pd.to_numeric(real_madrid_df[col])

# 5. Display the top 5 scorers for Real Madrid this season
top_madrid_scorers = real_madrid_df.sort_values(by='goals', ascending=False).head(10)

print("--- Real Madrid Top 5 Scorers (2025/26) ---")
df_top_10 = pd.DataFrame(top_madrid_scorers[['player_name', 'goals', 'xG', 'assists', 'xA']])


# Make Rank a column
df_top_10 = df_top_10.reset_index(drop=True)
df_top_10 = df_top_10.reset_index()

df_top_10 = df_top_10.rename(columns={'index': 'Rank'})

df_top_10['Rank'] = df_top_10.index + 1

# Set the index to be invisible empty text
df_top_10




```
