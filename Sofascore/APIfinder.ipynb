```python
# import all the packages:
import json
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as chromeservice
from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
options = webdriver.ChromeOptions()
options.set_capability(
    "goog:loggingPrefs",
    {"performance":"ALL", "browser": "ALL"}
)
driver = webdriver.Chrome(options = options)
```

```python
driver.set_page_load_timeout(20)
try:
    driver.get('https://www.sofascore.com/football/match/benfica-real-madrid/Egbsgkb#id:15452752')
except:
    pass
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

time.sleep(3)

```

```python
logs_raw = driver.get_log("performance")
logs = [json.loads(lr['message'])['message'] for lr in logs_raw]
```

```python
for x in logs:
    if x['method'] == 'Network.responseReceived':
        url = x['params']['response']['url']
        if 'shotmap' in url:
            print(url)
            break
```

```python
# Final Sofascore scrapper - this just works fine
import requests
import json

headers = {"User-Agent": "Mozilla/5.0"}

def get_sofascore_data(match_id, endpoint):
    url = f"https://www.sofascore.com/api/v1/event/{match_id}/{endpoint}"
    response = requests.get(url, headers=headers)
    return response.json()

# get shotmap
shotmap = get_sofascore_data("15452752", "shotmap")
print(json.dumps(shotmap, indent=2,ensure_ascii=False))
```
