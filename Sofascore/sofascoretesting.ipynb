```python
# Sofascore scrapper - this just works fine
import requests
import json

headers = {"User-Agent": "Mozilla/5.0"}

def get_sofascore_data(match_id, endpoint):
    url = f"https://www.sofascore.com/api/v1/event/{match_id}/{endpoint}"
    response = requests.get(url, headers=headers)
    return response.json()

# get shotmap
shotmap = get_sofascore_data("14083617", "statistics")
print(json.dumps(shotmap, indent=2,ensure_ascii=False))
```
